```python
import pandas as pd

df = pd.read_csv("social-media.csv")

col = "Avg_Daily_Usage_Hours"

Q1 = df[col].quantile(0.25)
Q3 = df[col].quantile(0.75)

IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound, "Upper Bound:", upper_bound)

df_no_outliers = df[(df[col] >= lower_bound) & (df[col] <= upper_bound)]

print("Original shape:", df.shape)
print("After removing outliers:", df_no_outliers.shape)
```

```python
df["Affects_Academic_Performance"] = df["Affects_Academic_Performance"].map({ "Yes": 1, "No": 0 })
```

```python
df = pd.get_dummies(df, drop_first=True)
df
```

```python
df.drop(columns=["Student_ID"])
df
```

```python
x = df.drop(columns=["Addicted_Score", "Mental_Health_Score"])
y_add = df["Addicted_Score"]
y_mh = df["Mental_Health_Score"]
```

```python
from sklearn.model_selection import train_test_split

x_train_add, x_test_add, y_train_add, y_test_add = train_test_split(x, y_add, test_size=0.2, random_state=42)

x_train_mh, x_test_mh, y_train_mh, y_test_mh = train_test_split(x, y_mh, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression

lr_add = LinearRegression()
lr_add.fit(x_train_add, y_train_add)

lr_mh = LinearRegression()
lr_mh.fit(x_train_mh, y_train_mh)
```

```python
y_pred_add = lr_add.predict(x_test_add)

print("Prediction of Addiction score is done")
```

```python
lr_mh = LinearRegression()
lr_mh.fit(x_train_mh, y_train_mh)

y_pred_mh = lr_mh.predict(x_test_mh)
print("Prediction of mental health score is done")
```

```python
from sklearn.metrics import r2_score, mean_absolute_error

print("Addiction Score")
print("R2 Score:", r2_score(y_test_add, y_pred_add))
print("MAE:", mean_absolute_error(y_test_add, y_pred_add))

print("\nMental Health Score")
print("R2 Score:", r2_score(y_test_mh, y_pred_mh))
print("MAE:", mean_absolute_error(y_test_mh, y_pred_mh))
```

```python
from sklearn.metrics import accuracy_score

# Convert actual & predicted values to classes
y_test_class = (y_test_add >= 7).astype(int)
y_pred_class = (y_pred_add >= 7).astype(int)

accuracy = accuracy_score(y_test_class, y_pred_class)

print("Accuracy:", accuracy)
```

```python
from sklearn.metrics import precision_score
precision = precision_score(y_test_class, y_pred_class)

print("Precision:", precision)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load the CSV file
df = pd.read_csv("social-media.csv")

# --- Scatter Plot 1: Addiction Score vs Usage Hours ---
x1 = df["Avg_Daily_Usage_Hours"]
y1 = df["Addicted_Score"]

# Fit a line (linear regression)
m1, b1 = np.polyfit(x1, y1, 1)

# Calculate correlation coefficient (R²)
r1 = np.corrcoef(x1, y1)[0,1]**2

plt.figure(figsize=(8,6))
plt.scatter(x1, y1, c="blue", alpha=0.6, edgecolors="k", label="Data Points")
plt.plot(x1, m1*x1 + b1, color="red", linewidth=2, label=f"Best Fit Line (R²={r1:.2f})")
plt.xlabel("Average Daily Usage (Hours)")
plt.ylabel("Addiction Score")
plt.title("Usage Hours vs Addiction Score with Best Fit Line")
plt.legend()
plt.grid(True)
plt.show()

# --- Scatter Plot 2: Mental Health Score vs Usage Hours ---
x2 = df["Avg_Daily_Usage_Hours"]
y2 = df["Mental_Health_Score"]

# Fit a line
m2, b2 = np.polyfit(x2, y2, 1)

# Calculate correlation coefficient (R²)
r2 = np.corrcoef(x2, y2)[0,1]**2

plt.figure(figsize=(8,6))
plt.scatter(x2, y2, c="green", alpha=0.6, edgecolors="k", label="Data Points")
plt.plot(x2, m2*x2 + b2, color="red", linewidth=2, label=f"Best Fit Line (R²={r2:.2f})")
plt.xlabel("Average Daily Usage (Hours)")
plt.ylabel("Mental Health Score")
plt.title("Usage Hours vs Mental Health Score with Best Fit Line")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import joblib

# Load dataset
df = pd.read_csv("social-media.csv")

# Convert Yes/No → 1/0
df["Affects_Academic_Performance"] = df["Affects_Academic_Performance"].map({
    "Yes": 1,
    "No": 0
})

# One-hot encoding
df = pd.get_dummies(df, drop_first=True)

# Features & targets
X = df.drop(columns=["Addicted_Score", "Mental_Health_Score", "Student_ID"])
y_add = df["Addicted_Score"]
y_mh = df["Mental_Health_Score"]

# Split
X_train_add, X_test_add, y_train_add, y_test_add = train_test_split(
    X, y_add, test_size=0.2, random_state=42
)

X_train_mh, X_test_mh, y_train_mh, y_test_mh = train_test_split(
    X, y_mh, test_size=0.2, random_state=42
)

# Train models
lr_add = LinearRegression()
lr_add.fit(X_train_add, y_train_add)

lr_mh = LinearRegression()
lr_mh.fit(X_train_mh, y_train_mh)

# Save models
joblib.dump(lr_add, "addiction_model.pkl")
print("Addiction model saved")

joblib.dump(lr_mh, "mental_health_model.pkl")
print("Mental Health model saved")
```
